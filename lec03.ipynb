```python
# Yeh ek simple function hai jo do numbers ko add karta hai.
def add(a, b):
  return a + b

# Yahan hum 'add' function ko call kar rahe hain (5 aur 5 ke saath) aur result print kar rahe hain.
print(add(5, 5))

# Yeh ek 'lambda function' hai, jo ek chhota anonymous function hota hai.
# Ismein teen arguments (a, b, c) hain.
# Yeh 'b * a + c' ka calculation karta hai.
x = lambda a, b, c: b * a + c

# Yahan hum 'x' lambda function ko call kar rahe hain (2, 4, aur 3 ke saath) aur result print kar rahe hain.
# Calculation: (4 * 2) + 3 = 8 + 3 = 11
print(x(2, 4, 3))
```

```python
# Ek empty string (khali string) banayi hai 'st' naam se.
st = ""
# 'st' string ki length print kar rahe hain, jo ki abhi 0 hai.
print(len(st))

# 'st' ko 'anand yadav' value de rahe hain.
st = "anand yadav"

# 'st' string ko uppercase mein convert karke print kar rahe hain (ANAND YADAV).
print(st.upper())

# 'st' string ke har word ka pehla letter capital karke print kar rahe hain (Anand Yadav).
print(st.title())

# 'st' string ka sirf pehla letter capital karke print kar rahe hain (Anand yadav).
print(st.capitalize())

# 'st' string mein 'a' kitni baar aata hai, woh count karke print kar rahe hain.
print(st.count('a'))

# 'st' string ke characters ko alphabetically sort karke list ke roop mein print kar rahe hain.
print(sorted(st))
# Original 'st' string ko print kar rahe hain.
print(st)

# 'st' string se leading aur trailing whitespace (agar ho toh) remove karke 's' mein store kar rahe hain.
# Iss case mein koi whitespace nahi hai, toh 's' ki value 'st' jaisi hi rahegi.
s = st.strip()

# 's' string ko print kar rahe hain.
print(s)
```

```python
# Yahan teen string variables ko initialize kar rahe hain.
s = "today"
b = "    weather  " # Ismein shuru aur end mein extra spaces hain.
c = "is GooD  "  # Ismein bhi extra spaces hain aur 'GooD' mein capitalization galat hai.

# Har string se shuru aur end ke extra spaces (whitespace) hata rahe hain.
s = s.strip() # 'today'
b = b.strip() # 'weather'
c = c.strip() # 'is GooD'

# Har string ke har word ka pehla letter capital kar rahe hain.
s = s.title() # 'Today'
b = b.title() # 'Weather'
c = c.title() # 'Is Good'

# String 'c' mein 'Good' ko 'cool' se badal rahe hain.
c = c.replace('Good', 'cool') # 'Is cool'

# Teeno strings ko ek space ke saath jodkar print kar rahe hain.
# Output: 'Today Weather Is cool'
print(s + ' ' + b + ' ' + c)

# String 'c' mein 'l' character ki pehli occurrence ka index find kar rahe hain.
# 'Is cool' mein 'l' ka index 6 hai (0-based indexing).
print(c.find('l'))

# Check kar rahe hain ki string 'c' 'Is' se shuru hoti hai ya nahi.
# Output: True, kyunki 'Is cool' 'Is' se hi shuru hota hai.
print(c.startswith('Is'))
```

```python
# Ek string ko initialize kiya hai, jo ki ek palindrome (ulta seedha ek jaisa) hai.
s = "A man, a plan, a canal: Panama"

# String ko sabhi lowercase letters mein convert kar rahe hain.
# Ab 's' ban jayega "a man, a plan, a canal: panama".
s = s.lower()

# String se shuru aur ant ke faltu spaces (whitespace) hata rahe hain.
# Is case mein shuru ya ant mein koi space nahi hai, toh string waisi hi rahegi.
s = s.strip()

# Pichhle code mein `s.remove()` galat tha, kyunki Python strings mein `remove()` method nahi hota.
# Strings immutable hoti hain aur characters ko hataane ke liye naya string banana padta hai.
# Yahan hum ek naya string bana rahe hain jismein sirf alphanumeric characters (letters aur numbers) honge.
# Isse punctuation aur spaces hat jayenge.
s_cleaned = "".join(char for char in s if char.isalnum())

# Clean ki hui string ko print kar rahe hain.
# Output hoga: "amanaplanacanalpanama"
print(s_cleaned)
```

```python
# Ek dictionary (key-value pairs ka collection) banayi hai 'dct' naam se.
dct = {
    "name": ["anand", "virat"], # 'name' key mein ek list hai.
    "age": 19,
    "class": 10,
    "roll no": 11,
}

# Dictionary mein ek naya key-value pair 'id': 12 add kar rahe hain.
dct['id'] = 12

# Dictionary se 'name' key aur uski value ko delete kar rahe hain.
del dct["name"]

# Modified dictionary ko print kar rahe hain.
print(dct)

# Yeh line comment ki hui hai, agar uncomment karenge toh 'name' key delete hone ke baad error degi.
# print(dct['name'][0])
```
