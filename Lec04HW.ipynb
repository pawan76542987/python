```python
# This script collects weekly sales data for various items.
# It prompts the user to enter item names and quantities sold until 'end' is entered.
# Finally, it displays a summary of the total sales for each item.

# Initialize an empty dictionary to store weekly sales data.
# The keys will be item names (strings) and values will be total quantities sold (integers).
weekly_sales = {}

# Loop indefinitely to collect sales data until the user types 'end'.
while True:
    # Prompt the user to enter an item name.
    # .strip() removes leading/trailing whitespace, and .lower() converts to lowercase for consistent keying.
    item = input("Enter item name (or type 'end' to finish): ").strip().lower()

    # Check if the user wants to end the input process.
    if item == "end":
        print("\n--- Sales Data Entry Complete ---")
        break

    try:
        # Prompt the user for the quantity sold for the current item.
        # int() converts the input string to an integer.
        quantity = int(input(f"How many {item}s were sold?: "))

        # Ensure the quantity is not negative.
        if quantity < 0:
            print("Quantity cannot be negative. Please enter a positive number.")
            continue

        # Update the weekly_sales dictionary.
        # If the item already exists, add the new quantity to its current total.
        # If the item is new, add it to the dictionary with the given quantity.
        if item in weekly_sales:
            weekly_sales[item] += quantity
        else:
            weekly_sales[item] = quantity
        print(f"Recorded: +{quantity} {item}(s)\n")

    except ValueError:
        # Handle cases where the user enters non-integer input for quantity.
        print("Invalid input for quantity. Please enter a whole number.\n")
    except Exception as e:
        # Catch any other unexpected errors during input.
        print(f"An unexpected error occurred: {e}\n")


# After the input loop, check if any sales data was recorded.
if not weekly_sales:
    print("No items were sold this week.")
else:
    # Print a header for the sales summary.
    print("\n--- Weekly Sales Summary ---")
    # Iterate through the collected weekly_sales dictionary.
    # For each item, print its name (capitalized for readability) and the total quantity sold.
    for item, total_sold in weekly_sales.items():
        print(f"• {item.capitalize()}: {total_sold} sold")
    print("----------------------------")
```
