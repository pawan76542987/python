```python
import pandas as pd

df1 = pd.DataFrame({
    'id':[1,2,3,4],
    'name':['raja', 'rani', 'rahul', 'rhea'],
    'age':[25,30,35,40]
})

df2 = pd.DataFrame({
    'id':[3,4,5,6],
    'name':['rahul', 'rhea', 'ravi', 'reena'],
    'age':[35,40,45,50]
})

# Inner Join: Returns only the rows where there is a match in both DataFrames based on the 'id' column.
# It combines rows that have the same 'id' in both df1 and df2.
df3 = pd.merge(df1, df2, on='id',how = 'inner')
print("\nInner Join Results:")
print(df3)

# Left Join: Returns all rows from the left DataFrame (df1), and the matched rows from the right DataFrame (df2).
# If there's no match in df2, the columns from df2 will be filled with NaN.
df3 = pd.merge(df1, df2, on='id',how = 'left')
print("\nLeft Join Results:")
print(df3)

# Right Join: Returns all rows from the right DataFrame (df2), and the matched rows from the left DataFrame (df1).
# If there's no match in df1, the columns from df1 will be filled with NaN.
df3 = pd.merge(df1, df2, on='id',how = 'right')
print("\nRight Join Results:")
print(df3)

# Outer Join: Returns all rows when there is a match in one of the DataFrames.
# It includes all rows from both DataFrames, filling NaN where there is no match.
df3 = pd.merge(df1, df2, on='id',how = 'outer')
print("\nOuter Join Results:")
print(df3)

# Cross Join: Returns the Cartesian product of the two DataFrames.
# Every row from df1 is combined with every row from df2.
df3 = pd.merge(df1, df2,how = 'cross')
print("\nCross Join Results:")
print(df3)
```
