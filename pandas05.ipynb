```python
# Import the pandas library, which is essential for working with DataFrames
import pandas as pd

# Create a dictionary named 'dct' to store data
# This dictionary contains three lists: 'name', 'age', and 'roll'
# Each list represents a column of data
dct = {
    'name' : ['tom','nick','krish','jack'], # List of names
    'age' : [20,21,19,18],          # List of ages
    'roll' : [1,2,3,4]              # List of roll numbers
}

# Create a pandas DataFrame named 'df' from the 'dct' dictionary
# A DataFrame is a 2-dimensional labeled data structure with columns of potentially different types.
# It's similar to a spreadsheet or SQL table, or a dictionary of Series objects.
df = pd.DataFrame(dct)

# Create another pandas DataFrame named 'df1' using the same 'dct' dictionary
# This creates a separate copy of the DataFrame with the same data
df1 = pd.DataFrame(dct)

# The following lines are commented out as they are alternative ways to print/display DataFrames
# print(df) # This would print the DataFrame to the console
# df        # In a Colab notebook, just typing the DataFrame name on the last line displays it
# df1       # This would display df1 if it were the last line

# Display the 'df' DataFrame in a nicely formatted table in the Colab output
display(df)

# Display the 'df1' DataFrame in a nicely formatted table in the Colab output
display(df1)
```

```python
# Pandas DataFrame mein Data ko kaise access karte hain, iloc aur loc ka use karke

# --- Integer-location based indexing (iloc) ---
# 'iloc' ka use rows aur columns ko unki integer position (yaani index number) ke hisab se select karne ke liye hota hai.

# DataFrame ki pehli (first) row ko select aur display karte hain.
# 'iloc[0]' ka matlab hai 'integer index 0 par jo row hai, woh laao'.
# Isse ek pandas Series return hoti hai, jahan index column ke naam hote hain aur values us row ka data.
print("\n--- Pehli Row (iloc[0]) ---")
display(df.iloc[0]) # 'display' function output ko achhe se dikhati hai

# DataFrame ki pehli 5 rows ko select aur display karte hain.
# 'iloc[0:5]' ka matlab hai 'integer index 0 se shuru karke index 5 tak (lekin 5 ko shamil nahi karna) saari rows laao'.
# Isse ek naya DataFrame return hota hai jismein yeh rows hoti hain.
print("\n--- Pehli 5 Rows (iloc[0:5]) ---")
display(df.iloc[0:5])

# --- Label-based indexing (loc) ---
# 'loc' ka use rows aur columns ko unke labels (jaise column names, ya custom row labels) se select karne ke liye hota hai.
# Yeh data ko filter karne ke liye boolean conditions (True/False conditions) ko bhi support karta hai.

# Un sabhi rows ko select aur display karte hain jahan 'study_hours' column ki value 11 hai.
# Sabse pehle, `df['study_hours']==11` ek boolean Series banata hai (har row ke liye True ya False).
# Phir, `df.loc[...]` is boolean Series ka use karke sirf un rows ko select karta hai jahan condition True hai.
print("\n--- Rows jahan 'study_hours' 11 hai (loc with boolean indexing) ---")
display(df.loc[df['study_hours']==11])

# Un rows ko select aur display karte hain jahan 'study_hours' 11 hai, lekin sirf 'study_hours' aur 'sleep_hours' columns ko hi dikhate hain.
# Yahan, hum rows ko filter karne ke liye boolean indexing (`df['study_hours']==11`) ko columns ko select karne ke liye label-based indexing (`['study_hours','sleep_hours']`) ke saath combine kar rahe hain.
# `['study_hours','sleep_hours']` specifies karta hai ki filtered rows ke liye kaun se columns dikhane hain.
print("\n--- 'study_hours' aur 'sleep_hours' un rows ke liye jahan 'study_hours' 11 hai ---")
display(df.loc[df['study_hours']==11 , ['study_hours','sleep_hours']])

# Ek aur advanced example:
# Pehle hum 'study_hours' 11 wali rows lete hain aur unse sirf 'study_hours' aur 'sleep_hours' columns nikalte hain.
# Uske baad, `.iloc[4:8]` ka use karke, us chhote data set mein se integer index 4 se 8 (8 shamil nahi) tak ki rows ko select karte hain.
# Yeh dikhata hai ki aap 'loc' aur 'iloc' ko chain bhi kar sakte hain.
print("\n--- Chained loc aur iloc: study_hours=11 wali rows se kuch specific columns, aur phir unmein se kuch specific rows ---")
display(df.loc[df['study_hours']==11 , ['study_hours','sleep_hours']].iloc[4:8])

# Ab hum ek naya DataFrame 'df2' banate hain.
# Is 'df2' mein sirf woh rows hongi jahan 'study_hours' ki value 11 hai.
df2 = df.loc[df['study_hours']==11]
print("\n--- Naya DataFrame 'df2' (jahan study_hours = 11) ---")
display(df2)

# 'df2' DataFrame mein se sirf 'study_hours' aur 'sleep_hours' columns ko select karke dikhate hain.
# Yeh double square brackets `[['col1', 'col2']]` ka use karke multiple columns select karne ka tareeka hai.
print("\n--- 'df2' mein se sirf 'study_hours' aur 'sleep_hours' columns ---")
display(df2[['study_hours' , 'sleep_hours']])

# 'df2' DataFrame mein se integer index 4 se 8 (8 shamil nahi) tak ki rows ko select karte hain.
# Kyunki 'df2' ab chhota hai, iski 'iloc' indexing original 'df' se alag hogi.
print("\n--- 'df2' mein se integer index 4 se 8 tak ki rows ---")
display(df2.iloc[4:8])
```
